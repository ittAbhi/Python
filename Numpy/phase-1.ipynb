```python
import numpy as np
```

### Creating array from List

```python
# 1D Array (One-Dimensional Array)
# A 1D array stores elements in a single sequence.
arr_1d = np.array([1, 2, 3, 4, 5])
print("1D array: ", arr_1d)


# 2D Array (Two-Dimensional Array)
# A 2D array stores elements in rows and columns, similar to a matrix.

# Incorrect way to create a 2D array:
# arr_2d = np.array([1, 2, 3], [4, 5, 6])
# This gives a TypeError because np.array() expects the array data
# as its first argument. Both rows must be inside one outer list.

# Correct way to create a 2D array:
arr_2d = np.array([[1, 2, 3], [4, 5, 6]])
print("2D array: ", arr_2d)
```

##List vs Numpy array

```python

import numpy as np
import time

# PART 1: Python list multiplication

# Create a normal Python list containing three elements.
py_list = [1, 2, 3]

# Multiplying a Python list by 2 repeats the entire list twice.
# It does NOT multiply each element by 2.
print("Python list multiplication:", py_list * 2)
# Output: [1, 2, 3, 1, 2, 3]


# PART 2: NumPy array multiplication

# Create a NumPy array containing three elements.
np_array = np.array([1, 2, 3])

# Multiplying a NumPy array by 2 multiplies each element by 2.
# This is called element-wise multiplication.
print("NumPy array multiplication:", np_array * 2)
# Output: [2 4 6]


# PART 3: Measure Python list operation time

# Record the current time before starting the operation.
start = time.time()

# Create a list containing 1,000,000 elements.
# Each element is its index multiplied by 2.
py_list = [i * 2 for i in range(1000000)]

# Subtract the start time from the current time
# to calculate how many seconds the operation took.
print("\nList operation time:", time.time() - start)


# PART 4: Measure NumPy operation time

# Record the current time before starting the operation.
start = time.time()

# Generate numbers from 0 to 999,999 as a NumPy array.
# Multiply every element by 2 using a vectorized operation.
np_array = np.arange(1000000) * 2

# Calculate and print the elapsed time in seconds.
print("\nNumPy operation time:", time.time() - start)
```

## Creating array from scratch


```python

import numpy as np

# 1. np.zeros() - Create an array filled with zeros.
# (3, 4) means 3 rows and 4 columns.
# By default, the elements are floating-point numbers.
zeros = np.zeros((3, 4))
print("Zeros array:\n", zeros)


# 2. np.ones() - Create an array filled with ones.
# (3, 4) means 3 rows and 4 columns.
# By default, the elements are floating-point numbers.
ones = np.ones((3, 4))
print("Ones array:\n", ones)


# 3. np.full() - Create an array filled with a specific value.
# (2, 2) means 2 rows and 2 columns.
# 7 is the value assigned to every element.
full = np.full((2, 2), 7)
print("Full array:\n", full)


# 4. np.random.random() - Create an array of random numbers.
# (2, 3) means 2 rows and 3 columns.
# Generates random floating-point numbers from 0 (inclusive)
# to 1 (exclusive).
# The values may change each time we run the program.
random = np.random.random((2, 3))
print("Random array:\n", random)


# 5. np.arange() - Create an array with a sequence of numbers.
# Syntax: np.arange(start, stop, step)
# start = 0  -> Starting number (included).
# stop  = 11 -> Ending limit (excluded).
# step  = 2  -> Difference between consecutive numbers.
sequence = np.arange(0, 11, 2)
print("Sequence array:\n", sequence)
```

### Vector, Matrix, Tensor

```python
import numpy as np  # Import NumPy library and give it the short name 'np'

# 1. VECTOR (1D ARRAY)
# A vector is a one-dimensional array.
# It contains elements arranged in a single sequence.
vector = np.array([1, 2, 3])

# Print the vector
print("Vector: ", vector)


# 2. MATRIX (2D ARRAY)
# A matrix is a two-dimensional array.
# It contains rows and columns.
# Here, we have 2 rows and 3 columns.
matrix = np.array([
    [1, 2, 3],  # First row
    [4, 5, 6]   # Second row
])

# Print the matrix
print("Matrix: ", matrix)


# 3. TENSOR (3D ARRAY)
# A tensor is a general term for a multidimensional array.
# Here, we have a 3D array containing 2 blocks.
# Each block has 2 rows and 2 columns.
tensor = np.array([
    [[1, 2], [3, 4]],  # First block (2 rows, 2 columns)
    [[5, 6], [7, 8]]   # Second block (2 rows, 2 columns)
])

# Print the tensor
print("Tensor: ", tensor)
```

### Array Properties


```python

import numpy as np

# Create a 2D NumPy array with 2 rows and 3 columns.
arr = np.array([
    [1, 2, 3],  # First row
    [4, 5, 6]   # Second row
])

# shape: Returns the number of elements along each dimension.
# Here, there are 2 rows and 3 columns.
print("Shape:", arr.shape)

# ndim: Returns the number of dimensions of the array.
# This is a 2D array, so the answer is 2.
print("Dimension:", arr.ndim)

# size: Returns the total number of elements in the array.
# 2 rows * 3 columns = 6 elements.
print("Size:", arr.size)

# dtype: Returns the data type of the elements.
# NumPy usually stores these integers as int32 or int64,
# depending on the platform and configuration.
print("DType:", arr.dtype)
```

## Array Reshaping


```python

import numpy as np

# Create a 1D array containing numbers from 0 to 11.
# np.arange(12) generates numbers from 0 up to (but not including) 12.
arr = np.arange(12)
print("Original array:", arr)


# RESHAPE
# Convert the 1D array into a 2D array with 3 rows and 4 columns.
# The total number of elements must remain the same: 3 * 4 = 12.
reShaped = arr.reshape((3, 4))
print("\nReshaped array:", reShaped)


# FLATTEN
# Convert the 2D array back into a 1D array.
# flatten() returns a COPY, so changes to it do not affect the original.
flattened = reShaped.flatten()
print("\nFlattened array:", flattened)


# RAVEL
# Convert the 2D array into a 1D array.
# ravel() returns a VIEW when possible, so it can share data
# with the original array instead of creating a separate copy.
raveled = reShaped.ravel()
print("\nRaveled array:", raveled)


# TRANSPOSE
# Swap the rows and columns of the 2D array.
# The original shape is (3, 4), and the transposed shape is (4, 3).
transpose = reShaped.T
print("\nTransposed array:", transpose)
```
