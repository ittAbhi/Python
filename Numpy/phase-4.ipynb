```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Creating NumPy Arrays

# Create a 2D array with 2 rows and 3 columns.
# np.array() converts a nested list into a NumPy array.
array1 = np.array([[1, 2, 3],
                   [4, 5, 6]])


# Create a 3x3 array filled with random floating-point values.
# np.random.rand() generates random values in the range [0, 1).
array2 = np.random.rand(3, 3)


# Create a 4x4 array filled with zeros.
# np.zeros() initializes every element to 0.
array3 = np.zeros((4, 4))
```

```python
# Saving NumPy Arrays to Files
# np.save() saves a NumPy array to a binary file with the .npy extension.
# The file preserves the array's data and shape for later use.

np.save('array1.npy', array1)
np.save('array2.npy', array2)
np.save('array3.npy', array3)
```

```python
# Loading a NumPy Array from a File
# np.load() reads a saved .npy file and restores the NumPy array.
# The file must exist at the specified path.

loaded_array1 = np.load('array1.npy')

# Display the loaded array.
print(loaded_array1)
```

```python
# Loading and Displaying an Image Using NumPy and Matplotlib

try:
    # Load the image data from a .npy file.
    # The file must contain image data stored as a NumPy array.
    logo = np.load('numpy-logo.npy')

    # Create a figure to display both images side by side.
    plt.figure(figsize=(10, 5))

    # Create the first subplot (1 row, 2 columns, position 1).
    plt.subplot(121)
    plt.imshow(logo)  # Display the original image.
    plt.title("NumPy Logo")
    plt.grid(False)  # Hide grid lines for a cleaner image.

    # Invert the image values to create a dark/inverted version.
    # For normalized image data in the range [0, 1],
    # subtracting each value from 1 inverts its intensity.
    dark_logo = 1 - logo

    # Create the second subplot (1 row, 2 columns, position 2).
    plt.subplot(122)
    plt.imshow(dark_logo)  # Display the inverted image.
    plt.title("NumPy Dark Logo")
    plt.grid(False)

except FileNotFoundError:
    # Handle the error if the specified file does not exist.
    print("NumPy logo file not found")
```
