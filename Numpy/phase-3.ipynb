# Advance operation with Business examples

```python

import numpy as np
import matplotlib.pyplot as plt
```

```python
# Data structure: [restaurant_id, 2021, 2022, 2023, 2024]
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

# Display the title of the analysis.
print("==== Zomato sales analysis ====")

# Check the shape of the dataset.
# shape returns a tuple: (number of rows, number of columns).
# Here, rows represent restaurants and columns represent IDs and yearly sales.
print("\nSales data shape:", sales_data.shape)

# Slicing: Select the first 3 rows (first 3 restaurants).
# The start index is included, while the stop index is excluded.
print("\nSample data for the first 3 restaurants:\n", sales_data[:3])
```

```python
# Total sales per year
# axis=0 calculates the sum column-wise (down each column).
# The first column contains restaurant IDs, so exclude it when calculating sales totals.
print(np.sum(sales_data, axis=0))

# Select all rows and columns from index 1 onward, excluding restaurant IDs.
# axis=0 sums sales across all restaurants for each year.
yearly_total = np.sum(sales_data[:, 1:], axis=0)
print("Yearly total: ", yearly_total)


# Minimum sales per restaurant
# axis=1 calculates the minimum row-wise (across each restaurant's yearly sales).
# Each result represents one restaurant's lowest annual sales.
min_sales = np.min(sales_data[:, 1:], axis=1)
print("Minimum sales: ", min_sales)


# Maximum sales per year
# axis=0 finds the highest sales value among all restaurants for each year.
max_sales = np.max(sales_data[:, 1:], axis=0)
print("Maximum sales: ", max_sales)


# Average sales per restaurant
# axis=1 calculates the mean of each restaurant's sales from 2021 to 2024.
avg_sales = np.mean(sales_data[:, 1:], axis=1)
print("Average sales: ", avg_sales)
```

```python
# Cumulative Sales
# np.cumsum() calculates the cumulative sum (running total) of array elements.
# sales_data[:, 1:] selects all restaurants and excludes the restaurant ID column.
# axis=1 calculates the cumulative sum row-wise, across each restaurant's yearly sales.
cumsum = np.cumsum(sales_data[:, 1:], axis=1)
print(cumsum)


# Plot the average cumulative sales across all restaurants.
plt.figure(figsize=(8, 6))  # Set the figure size (width, height) in inches.

# Calculate the average cumulative sales for each year across all restaurants.
# axis=0 calculates the mean column-wise.
plt.plot(np.mean(cumsum, axis=0))

# Add a title and labels to describe the chart.
plt.title("Average Cumulative Sales Across All Restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")

# Display grid lines to make values easier to compare.
plt.grid(True)

# Display the plot.
plt.show()
```

```python
# Vector Operations
# A vector is a one-dimensional array containing numerical values.

vector1 = np.array([1, 2, 3, 4, 5])
vector2 = np.array([6, 7, 8, 9, 10])


# Vector Addition
# Adds corresponding elements of both vectors.
print("Vector addition: ", vector1 + vector2)


# Element-wise Multiplication
# Multiplies corresponding elements of both vectors.
print("\nMultiplication vector: ", vector1 * vector2)


# Dot Product
# Multiplies corresponding elements and adds all the products.
# Both vectors must have the same number of elements.
print("\nDot Product: ", np.dot(vector1, vector2))


# Angle Between Two Vectors
# Formula: cos(theta) = (A · B) / (||A|| * ||B||)
# np.dot() calculates the dot product.
# np.linalg.norm() calculates the magnitude (length) of a vector.
# np.arccos() converts the cosine value into an angle in radians.
angle = np.arccos(
    np.dot(vector1, vector2) /
    (np.linalg.norm(vector1) * np.linalg.norm(vector2))
)

print("Angle (radians): ", angle)
```

```python
# Vectorization in NumPy
# np.vectorize() applies a function to each element of an array.
# It allows a scalar function to operate on array elements.

restaurant_types = np.array(['biryani', 'chinese', 'pizza', 'burger', 'cafe'])

# Create a vectorized version of str.upper().
# str.upper() converts a string to uppercase.
vectorized_upper = np.vectorize(str.upper)

# Apply the function to every element in the array.
print("Vectorized Upper: ", vectorized_upper(restaurant_types))
```

```python
# Calculating Average Monthly Sales
# sales_data[:, 1:] selects all restaurants and excludes the restaurant ID column.
# Divide annual sales by 12 to estimate the average monthly sales for each year.
# NumPy broadcasting applies the division element-wise to every selected value.
monthly_avg = sales_data[:, 1:] / 12

print(monthly_avg)
```
